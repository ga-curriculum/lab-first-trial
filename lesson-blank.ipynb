## Problem 1: BMI

You're a physician compute the body mass index (BMI) of one of your patients.

**Step 1:** Create a variable `weight` that has the value 90.2, the patient's weight in kilograms. Then, create a variable called `height` with a value of 1.8, the patient's height in meters.

```python
weight = 90.2
height = 1.8
```

**Step 2:** Compute the patient's BMI as follows:

$$BMI = \frac{w}{h^2}$$

Define this value as a variable called `bmi` and print this variable to the screen.

```python
bmi = weight / height**2
print(bmi)
```

## Problem 2: Movie Ratings
**Step 1:** Create a variable called `age` that represents a moviegoer's age.

```python
age = 16
```

**Step 2:** Write a conditional statement that tells the user what movies they can see. The logic should work as follows:

* If the user is 12 or below, your code should print "You may see movies rated G or PG"
* If the user is between 13 and 16 years old, your code should print "You may see movies rated PG-13 or below"
* If the user is 17 or older, your code should print "You can see any movie without parental guidance!"

```python
if age <= 12:
    print("You may see movied rated G or PG")
elif age <= 16:
    print("You may see movies rated PG-13 or below")
else:
    print("You can see any movie without parental guidance!")
```

## Problem 3: Rapid Loops

**Loop 3:** Print only the elements of the given list that are divisible by 3.

```python
my_list = [56, 93, 67, 66, 34, 42, 75, 79, 101, 45]
```

```python
for item in my_list:
    if item % 3 == 0:
        print(item)
```
